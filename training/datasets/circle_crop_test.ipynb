```python
# Generate circle mask
from experiment.datasets.circle_crop import circle_crop, circle_mask
from torchvision.transforms import functional as F
import torch
import torchvision.transforms as T
from PIL import Image, ImageDraw
import numpy as np

for i in range(10):
    image = train_dataset[i][0]

    mask = circle_mask(image.shape[1:])

    # Apply circular crop
    cropped_tensor = circle_crop(image, mask)

    angle = 45  # For demonstration, rotate by 45 degrees.
    rotated_tensor = F.rotate(T.ToPILImage()(cropped_tensor), angle, interpolation=Image.BILINEAR)
    rotated_tensor = T.ToTensor()(rotated_tensor)

    # Convert tensor back to image and display
    to_image = T.ToPILImage()
    rotated_image = to_image(rotated_tensor)
    #rotated_image.show()
    rotated_image.save(f"experiment/datasets/output/rotated_image_{i}.png")
```
