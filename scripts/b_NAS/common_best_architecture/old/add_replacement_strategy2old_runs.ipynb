```python
import wandb
import re
api = wandb.Api()
```

```python
run = api.run("ga92xug/SL-NAS-common-best/7ir6f1en")
run.config["model"]["replacement_group_strategy"] = "strategy_cifar10"
print(run.config)
run.update()
```

```python
map_strategies_2_replacement_groups = {
    "cifar10": "cifar10",
    "cifar10_rot": "mnist_rot",
    "galaxy10": "galaxy10",
    "mnist": "cifar10",
    "mnist_rot": "mnist_rot",
    "ISIC_2019": "galaxy10"
}
```

```python
runs = api.runs('ga92xug/SL-NAS-common-best')
pattern = r'pure_strategy_.*_group_adjusted'
for run in runs:
    name = run.name
    if re.search(pattern, name):
        print(name)
        dataset_name = run.config["training"]["dataset"]["name"]
        if dataset_name == "Galaxy10_DECals":
            dataset_name = "galaxy10"
        replacement_group = map_strategies_2_replacement_groups[dataset_name]
        replacement_group_strategy = "strategy_" + replacement_group
        run.config["model"]["replacement_group_strategy"] = replacement_group_strategy
        run.update()
        print("dataset", dataset_name, "replacement", replacement_group_strategy)
        


```
