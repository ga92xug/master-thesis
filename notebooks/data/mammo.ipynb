```python
%load_ext autoreload
%autoreload 2
!cd ~/dev1/scaling-laws-ecnn

import pandas as pd
import numpy as np
import os
import matplotlib.pyplot as plt
import math

PATH = os.path.expanduser("~/Data/frischs/datasets/mammography/combined/")
```

```python
inbreast = os.path.join(PATH, "INbreast", "Benign Masses")
inbreast_all = [f for f in os.listdir(inbreast)]
inbreast_original = [f for f in os.listdir(inbreast) if "(" not in f]
len(inbreast_original)
```

```python
len(inbreast_all) / 72
```

```python

fig_size = 2.5*len(inbreast_original)/2
fig, ax = plt.subplots(int(math.ceil(len(inbreast_original)/2)), 2, figsize=(5, fig_size))

ax = ax.ravel()
for i, patient in enumerate(inbreast_original):
    ax[i].imshow(plt.imread(os.path.join(inbreast, patient)), cmap="gray")
    ax[i].axis("off")
    ax[i].set_title(patient)
```

```python
patient_0 = inbreast_all[0].split(" ")[0]
patients = [f for f in inbreast_all if patient_0 in f]

# we want to check out all the images for a single patient -> matplotlib subplot
fig_size = 2.5*len(patients)/2
fig, ax = plt.subplots(len(patients)//2, 2, figsize=(5, fig_size))

ax = ax.ravel()
for i, patient in enumerate(patients):
    ax[i].imshow(plt.imread(os.path.join(inbreast, patient)))
    ax[i].axis("off")
    ax[i].set_title(patient)

#plt.imshow(plt.imread(os.path.join(inbreast, inbreast_original[0])))
```
