```python
import matplotlib.pyplot as plt
import numpy as np

# Generate example accuracy data for a single run (replace this with your actual data)
num_epochs = 50
mean_accuracy = np.random.uniform(0.7, 0.9, num_epochs)  # Replace with your actual accuracy data
std_deviation = np.random.uniform(0.02, 0.05, num_epochs)  # Replace with your actual standard deviation data

# Create the line plot for the mean accuracy
plt.figure(figsize=(10, 6))
plt.plot(range(1, num_epochs + 1), mean_accuracy, label="Mean Accuracy", color="blue")

# Shade the area around the line to illustrate the standard deviation
plt.fill_between(
    range(1, num_epochs + 1),
    mean_accuracy - std_deviation,
    mean_accuracy + std_deviation,
    color="blue",
    alpha=0.3,
    label="Standard Deviation",
)

plt.title("Accuracy Over Epochs with Standard Deviation")
plt.xlabel("Epochs")
plt.ylabel("Accuracy")
plt.legend()
plt.grid(True)
plt.show()
```

```python
import numpy as np

# Define a function to convert data to NumPy arrays
def transform_data_to_arrays(downloaded_data, metric):
    transformed_data = {}  # Dictionary to store the transformed data

    # Iterate through labels and runs
    for label, runs_data in downloaded_data.items():
        transformed_data[label] = {}  # Initialize the label entry
        for run_id, run_data in runs_data.items():
            acc = np.array(run_data[metric])  # Convert the metric data to a NumPy array
            transformed_data[label][run_id] = acc  # Store the NumPy array

    return transformed_data

# Example usage:
downloaded_data = {
    "Label1": {
        "Run1": {"Metric1": [1.0, 2.0, 3.0]},
        "Run2": {"Metric1": [0.5, 1.5, 2.5]}
    },
    "Label2": {
        "Run1": {"Metric1": [0.8, 1.8, 2.8]},
        "Run2": {"Metric1": [0.7, 1.7, 2.7]}
    }
}

metric = "Metric1"

transformed_data = transform_data_to_arrays(downloaded_data, metric)

# Now you have transformed data as NumPy arrays, and you can easily calculate mean and std
for label, runs_data in transformed_data.items():
    for run_id, acc in runs_data.items():
        mean = np.mean(acc)
        std = np.std(acc)
        print(f"Label: {label}, Run ID: {run_id}, Mean: {mean}, Std: {std}")
```
